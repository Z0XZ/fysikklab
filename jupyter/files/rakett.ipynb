# Rakett med bevegelsesmengde

En rakett har ingenting å skyve mot i rommet. Den kommer seg fram fordi den skyter gass bakover, og da må raketten få like mye bevegelsesmengde framover. Massen til raketten minker hele tiden, så akselerasjonen er ikke konstant. Det er et godt problem for Python.

## Steg 1: Rakett i rommet

Raketten har startmassen 1000 kg, og 800 kg av det er drivstoff. Gassen blir skutt ut med farten u = 2500 m/s bakover i forhold til raketten. Vi skyter ut 1 kg om gangen og bruker at bevegelsesmengden er bevart i hvert lille «støt». Til slutt sammenligner vi med rakettlikningen Δv = u · ln(M₀/M), som man kan utlede med integrasjon.

```python
import numpy as np
import matplotlib.pyplot as plt

M = 1000.0      # startmasse i kg (rakett + drivstoff)
drivstoff = 800.0
u = 2500.0      # farten til gassen bakover, i forhold til raketten, i m/s
dm = 1.0        # masse som skytes ut i hvert steg, i kg

v = 0.0         # farten til raketten i m/s (i rommet, uten tyngdekraft)
masser, farter = [M], [v]

while drivstoff > 0:
    # Bevegelsesmengden før = etter:
    # M·v = (M − dm)·(v + dv) + dm·(v − u)   gir   dv = u·dm / (M − dm)
    dv = u * dm / (M - dm)
    v = v + dv
    M = M - dm
    drivstoff = drivstoff - dm
    masser.append(M)
    farter.append(v)

print(f"Sluttfart: {v:.0f} m/s med {M:.0f} kg igjen")
print(f"Rakettlikningen u·ln(M0/M): {u * np.log(1000 / M):.0f} m/s")

plt.plot(masser, farter)
plt.gca().invert_xaxis()          # massen minker mot høyre
plt.xlabel("massen til raketten / kg")
plt.ylabel("v / (m/s)")
plt.grid(True)
plt.show()
```

## Steg 2: Modellrakett som skytes rett opp

En modellrakett har 12 g drivstoff som brenner i 1,6 s. Skyvekraften er bevegelsesmengden gassen får per sekund: F = u · Δm/Δt. Nå virker også tyngden og luftmotstanden L = kv². Programmet kjører til raketten er på toppen. Tallene er et tenkt, men realistisk eksempel.

```python
import numpy as np
import matplotlib.pyplot as plt

g = 9.81
m_tom = 0.108          # kg, rakett uten drivstoff
drivstoff = 0.012      # kg
brennetid = 1.6        # s
u = 800.0              # utstrømningsfart i m/s
k = 2.2e-4             # luftmotstandstall i kg/m: L = k·v²

dmdt = drivstoff / brennetid      # kg/s
F_skyv = u * dmdt                 # skyvekraften: endringen i bevegelsesmengde per tid
print(f"Skyvekraft: {F_skyv:.1f} N")

t, h, v = 0.0, 0.0, 0.0
m = m_tom + drivstoff
dt = 0.001
tider, hoyder, farter = [], [], []

while v >= 0:                     # til raketten er på toppen
    F = -m * g - k * v * abs(v)   # tyngde og luftmotstand
    if t < brennetid:
        F = F + F_skyv
        m = m - dmdt * dt
    a = F / m
    v = v + a * dt
    h = h + v * dt
    t = t + dt
    tider.append(t)
    hoyder.append(h)
    farter.append(v)

print(f"Største fart: {max(farter):.1f} m/s")
print(f"Toppunkt: {h:.0f} m etter {t:.1f} s")

plt.subplot(2, 1, 1)
plt.plot(tider, hoyder)
plt.ylabel("h / m")
plt.grid(True)
plt.subplot(2, 1, 2)
plt.plot(tider, farter)
plt.xlabel("t / s")
plt.ylabel("v / (m/s)")
plt.grid(True)
plt.show()
```
