# Regresjon og linearisering i Python

## Steg 1: Tyngdeakselerasjonen fra en pendel

Svingetiden til en pendel er T = 2π√(L/g). Kvadrerer vi, får vi T² = (4π²/g) · L. Grafen av T² mot L er altså en rett linje gjennom origo med stigningstall 4π²/g. Pendeldataene under er simulert med g = 9,81 m/s² og litt tilfeldig målestøy. Bytt dem gjerne ut med deres egne målinger.

Med `cov=True` gir `np.polyfit` også usikkerheten i stigningstallet. Den relative usikkerheten i g blir den samme som i stigningstallet.

```python
import numpy as np
import matplotlib.pyplot as plt

# Simulerte målinger: svingetiden T for en pendel med lengde L
L = np.array([0.20, 0.40, 0.60, 0.80, 1.00, 1.20])        # m
T = np.array([0.901, 1.279, 1.558, 1.779, 2.017, 2.203])  # s

# Teorien sier T = 2π√(L/g), altså T² = (4π²/g) · L. Da skal T² mot L bli en rett linje.
koeff, kov = np.polyfit(L, T**2, 1, cov=True)
stigning, konstant = koeff
u_stigning = np.sqrt(kov[0, 0])    # standardusikkerheten i stigningstallet

g = 4 * np.pi**2 / stigning
u_g = g * u_stigning / stigning    # samme relative usikkerhet som stigningstallet

print(f"T² = {stigning:.3f} · L + {konstant:.3f}")
print(f"g = ({g:.2f} ± {u_g:.2f}) m/s²")

plt.plot(L, T**2, "o", label="målinger")
L_linje = np.linspace(0, 1.3, 50)
plt.plot(L_linje, stigning * L_linje + konstant, label="rett linje")
plt.xlabel("L / m")
plt.ylabel("T² / s²")
plt.legend()
plt.grid(True)
plt.show()
```

## Steg 2: Når vi ikke kjenner formelen

Anta at vi bare vet at T = C · L<sup>n</sup>, men ikke hva n er. Tar vi logaritmen, får vi ln T = ln C + n · ln L. Grafen av ln T mot ln L er da en rett linje med stigningstall n. Den bruker variablene fra stegene over, som cellene i en Jupyter-notatbok.

```python
# Hvis vi ikke kjenner formelen: prøv en potensfunksjon T = C · L^n.
# Da er ln T = ln C + n · ln L, altså en rett linje med stigningstall n.
n, lnC = np.polyfit(np.log(L), np.log(T), 1)
print(f"n = {n:.3f}")
print(f"C = {np.exp(lnC):.3f} s/m^n")

plt.loglog(L, T, "o")
plt.xlabel("L / m")
plt.ylabel("T / s")
plt.grid(True, which="both")
plt.show()
```

## Steg 3: Keplers tredje lov med ekte data

Keplers tredje lov sier at T² er proporsjonal med a³, der a er den store halvaksen i banen. Tallene under er kjente verdier for planetene, med a i astronomiske enheter (AU) og T i år. Dette er et nytt, selvstendig program.

```python
import numpy as np
import matplotlib.pyplot as plt

# Store halvakser a (AU) og omløpstider T (år) for planetene
planeter = ["Merkur", "Venus", "Jorda", "Mars", "Jupiter", "Saturn", "Uranus", "Neptun"]
a = np.array([0.387, 0.723, 1.000, 1.524, 5.203, 9.537, 19.19, 30.07])   # AU
T = np.array([0.241, 0.615, 1.000, 1.881, 11.86, 29.45, 84.02, 164.8])   # år

n, lnC = np.polyfit(np.log(a), np.log(T), 1)
print(f"T er proporsjonal med a^{n:.3f}")

for navn, ai, Ti in zip(planeter, a, T):
    print(f"{navn:8s} T²/a³ = {Ti**2 / ai**3:.3f} år²/AU³")

plt.loglog(a, T, "o")
for navn, ai, Ti in zip(planeter, a, T):
    plt.annotate(navn, (ai, Ti), textcoords="offset points", xytext=(5, -10))
plt.xlabel("a / AU")
plt.ylabel("T / år")
plt.grid(True, which="both")
plt.show()
```
