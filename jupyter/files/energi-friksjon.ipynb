# Energi med friksjon og luftmotstand

Uten friksjon blir all potensiell energi til kinetisk energi. Med friksjon og luftmotstand blir noe av energien til varme. Her simulerer vi bevegelsen med Eulers metode og fører regnskap over energien underveis. Går regnskapet opp?

## Steg 1: Kloss på et skråplan

En kloss på 2,0 kg glir 4,0 m ned et skråplan med helningsvinkel 25° og friksjonstall 0,20. Friksjonskraften er R = μN, og normalkraften er N = mg cos θ. Akselerasjonen er konstant, så vi kan sammenligne simuleringen med energibevaring: ½mv² = mgh − RL.

```python
import numpy as np
import matplotlib.pyplot as plt

g = 9.81            # m/s²
m = 2.0             # kg
vinkel = np.radians(25)
mu = 0.20           # friksjonstall
L = 4.0             # lengden på skråplanet i m
h0 = L * np.sin(vinkel)   # starthøyde i m

N = m * g * np.cos(vinkel)        # normalkraft
R = mu * N                        # friksjonskraft
a = g * np.sin(vinkel) - R / m    # akselerasjon langs planet

t, s, v = 0, 0, 0     # s, m, m/s
dt = 0.001            # s
tider, Ek, Ep, Q = [], [], [], []

while s < L:
    v = v + a * dt
    s = s + v * dt
    t = t + dt
    tider.append(t)
    Ek.append(0.5 * m * v**2)
    Ep.append(m * g * (h0 - s * np.sin(vinkel)))
    Q.append(R * s)               # friksjonsarbeidet blir til varme

print(f"Klossen er nederst etter {t:.2f} s med farten {v:.2f} m/s.")
v_energi = np.sqrt(2 * (m * g * h0 - R * L) / m)
print(f"Energibevaring gir {v_energi:.2f} m/s.")
print(f"Varme fra friksjonen: {R * L:.1f} J av {m * g * h0:.1f} J")
```

## Steg 2: Energien underveis

Vi plotter den kinetiske energien, den potensielle energien og varmen fra friksjonen mot tiden. Den stiplede linja er summen av de tre. Koden bruker listene fra steg 1.

```python
tider = np.array(tider)
Ek, Ep, Q = np.array(Ek), np.array(Ep), np.array(Q)

plt.plot(tider, Ek, label="kinetisk energi")
plt.plot(tider, Ep, label="potensiell energi")
plt.plot(tider, Q, label="varme (friksjonsarbeid)")
plt.plot(tider, Ek + Ep + Q, "k--", label="sum")
plt.xlabel("t / s")
plt.ylabel("energi / J")
plt.legend()
plt.grid(True)
plt.show()
```

## Steg 3: Skiløper med luftmotstand

En skiløper på 70 kg kjører 300 m rett ned en bakke med helning 15°. Luftmotstanden er L = kv², så den øker med farten, og akselerasjonen er ikke konstant. Da kan vi ikke bruke formlene for konstant akselerasjon, men simuleringen virker like godt. Vi summerer arbeidet fra friksjonen og luftmotstanden: I hvert tidssteg er arbeidet kraften ganger strekningen vΔt.

```python
import numpy as np
import matplotlib.pyplot as plt

g = 9.81
m = 70.0                 # kg, skiløper med utstyr
vinkel = np.radians(15)
mu = 0.05                # friksjonstall mellom ski og snø
k = 0.25                 # luftmotstandstall i kg/m: L = k·v²
lengde = 300.0           # m nedover bakken

R = mu * m * g * np.cos(vinkel)
t, s, v = 0, 0, 0
dt = 0.01
tider, farter = [], []
W_friksjon, W_luft = 0, 0

while s < lengde:
    L = k * v**2                                  # luftmotstanden endrer seg med farten
    a = g * np.sin(vinkel) - (R + L) / m
    v = v + a * dt
    s = s + v * dt
    t = t + dt
    W_friksjon = W_friksjon + R * v * dt          # arbeid = kraft · strekning
    W_luft = W_luft + L * v * dt
    tider.append(t)
    farter.append(v)

Ep_tap = m * g * lengde * np.sin(vinkel)
Ek = 0.5 * m * v**2
print(f"Etter {lengde:.0f} m: t = {t:.1f} s og v = {v:.1f} m/s ({v * 3.6:.0f} km/h)")
print(f"Tapt potensiell energi: {Ep_tap / 1000:.1f} kJ")
print(f"  kinetisk energi: {Ek / 1000:.1f} kJ")
print(f"  friksjon:        {W_friksjon / 1000:.1f} kJ")
print(f"  luftmotstand:    {W_luft / 1000:.1f} kJ")
print(f"  sum:             {(Ek + W_friksjon + W_luft) / 1000:.1f} kJ")

v_terminal = np.sqrt((m * g * np.sin(vinkel) - R) / k)
plt.plot(tider, farter)
plt.axhline(v_terminal, color="gray", linestyle="--", label="terminalfart")
plt.xlabel("t / s")
plt.ylabel("v / (m/s)")
plt.legend()
plt.grid(True)
plt.show()
```
