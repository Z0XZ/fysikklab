# Støt mellom vogner i Python

En rød vogn på 0,50 kg kjører inn i en blå vogn på 0,25 kg som står i ro. Farten til begge vognene er logget 40 ganger i sekundet. Bruk Python til å finne ut om bevegelsesmengden og den kinetiske energien er bevart.

## Data

Fila har tre kolonner: tid i sekunder, farten til den røde vogna og farten til den blå vogna i m/s. Den har desimalpunktum og komma som skilletegn. Dataene er laget med en modell av et støt med magnetstøtfangere, litt friksjon og målestøy, slik at de ligner målinger fra Smart Cart. De er ikke ekte målinger. Bytt gjerne ut fila med deres egne målinger.

## Steg 1: Les inn og plott

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt("vognstot_smartcart.csv", delimiter=",", skiprows=1)
t = data[:, 0]
v_rod = data[:, 1]
v_bla = data[:, 2]

m_rod = 0.50   # kg
m_bla = 0.25   # kg

plt.plot(t, v_rod, "r.-", label="rød vogn")
plt.plot(t, v_bla, "b.-", label="blå vogn")
plt.xlabel("t / s")
plt.ylabel("v / (m/s)")
plt.legend()
plt.grid()
plt.show()
```

## Steg 2: Bevegelsesmengde og kinetisk energi

```python
p_rod = m_rod * v_rod
p_bla = m_bla * v_bla
p_sum = p_rod + p_bla

ek_sum = 0.5 * m_rod * v_rod**2 + 0.5 * m_bla * v_bla**2

plt.plot(t, p_rod, "r", label="rød")
plt.plot(t, p_bla, "b", label="blå")
plt.plot(t, p_sum, "k--", label="sum")
plt.xlabel("t / s")
plt.ylabel("p / (kg m/s)")
plt.legend()
plt.grid()
plt.show()
```

## Steg 3: Gjennomsnitt før og etter støtet

```python
# Velg tidsrom der farten er omtrent konstant. Se på grafen og juster grensene.
foer = (t >= 0.5) & (t < 0.9)
etter = (t >= 1.2) & (t < 1.6)

p_for = np.mean(p_sum[foer])
p_etter = np.mean(p_sum[etter])
ek_for = np.mean(ek_sum[foer])
ek_etter = np.mean(ek_sum[etter])

print(f"p før:   {p_for:.4f} kg m/s")
print(f"p etter: {p_etter:.4f} kg m/s   ({p_etter / p_for * 100:.1f} %)")
print(f"Ek før:   {ek_for:.4f} J")
print(f"Ek etter: {ek_etter:.4f} J   ({ek_etter / ek_for * 100:.1f} %)")
```
