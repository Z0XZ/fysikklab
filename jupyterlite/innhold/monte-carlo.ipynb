# Usikkerhet med Monte Carlo-simulering

Regnereglene for usikkerhet er lette å bruke, men hvor kommer de fra, og når slutter de å virke? I en Monte Carlo-simulering lar vi datamaskinen gjøre det samme forsøket 100 000 ganger med tilfeldige målefeil, og ser hvor mye resultatet sprer seg.

## Steg 1: g fra en pendel

En pendel har lengden L = (0,800 ± 0,005) m og svingetiden T = (1,79 ± 0,02) s, og g = 4π²L/T². Vi trekker 100 000 tilfeldige lengder og svingetider fra en normalfordeling der standardavviket er usikkerheten, og regner ut g for hver. Standardavviket til alle g-verdiene er usikkerheten i g. Tallene endrer seg litt hver gang du kjører, fordi de er tilfeldige.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng()      # tilfeldige tall
n = 100_000                        # antall «forsøk» vi simulerer

# Målinger med usikkerhet (standardavvik)
L, dL = 0.800, 0.005    # m
T, dT = 1.79, 0.02      # s

L_trukket = rng.normal(L, dL, n)   # 100 000 tilfeldige lengder rundt 0,800 m
T_trukket = rng.normal(T, dT, n)
g = 4 * np.pi**2 * L_trukket / T_trukket**2

print(f"g = {g.mean():.2f} ± {g.std():.2f} m/s²  (Monte Carlo)")

# Sammenlign med reglene for usikkerhet
g0 = 4 * np.pi**2 * L / T**2
enkel = dL / L + 2 * dT / T
kvadrat = np.sqrt((dL / L)**2 + (2 * dT / T)**2)
print(f"g = {g0:.2f} m/s²")
print(f"Enkel regel:  ± {enkel * g0:.2f} m/s²  ({enkel * 100:.1f} %)")
print(f"Kvadratsum:   ± {kvadrat * g0:.2f} m/s²  ({kvadrat * 100:.1f} %)")

plt.hist(g, bins=80)
plt.xlabel("g / (m/s²)")
plt.ylabel("antall")
plt.show()
```

## Steg 2: Differansen mellom to nesten like tall

Vi finner massen av vannet i et beger ved å veie begeret med og uten vann. Begge veiingene har usikkerhet 0,5 g. Koden bruker `rng` og `n` fra steg 1.

```python
# Massen av vannet i et beger: full minus tom
m_full = rng.normal(152.3, 0.5, n)   # g
m_tom = rng.normal(148.9, 0.5, n)    # g
m_vann = m_full - m_tom

print(f"m = {m_vann.mean():.1f} ± {m_vann.std():.1f} g")
print(f"Relativ usikkerhet: {m_vann.std() / m_vann.mean() * 100:.0f} %")
print(f"Regelen for differanse: ± {0.5 + 0.5:.1f} g")
```

## Steg 3: Når regnereglene ikke holder

Rekkevidden til et skrått kast på flatt underlag er R = v² sin(2θ)/g. Ved 45° er sin(2θ) størst, så en liten feil i vinkelen gjør nesten ingen forskjell, og regnereglene sier at vinkelen ikke bidrar til usikkerheten. Men stemmer det?

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng()
n = 100_000
g = 9.81

# Rekkevidden til et skrått kast på flatt underlag: R = v² sin(2θ) / g
v = rng.normal(10.0, 0.2, n)                 # m/s
vinkel = np.radians(rng.normal(45, 3, n))    # 45 ± 3 grader
R = v**2 * np.sin(2 * vinkel) / g

R0 = 10.0**2 * np.sin(np.radians(90)) / g
print(f"Med v = 10,0 m/s og 45°: R = {R0:.2f} m")
print(f"Monte Carlo: gjennomsnitt {R.mean():.2f} m, standardavvik {R.std():.2f} m")
print(f"Andel kast som går lenger enn R0: {np.mean(R > R0) * 100:.0f} %")

plt.hist(R, bins=80)
plt.axvline(R0, color="k", linestyle="--")
plt.xlabel("R / m")
plt.ylabel("antall")
plt.show()
```
